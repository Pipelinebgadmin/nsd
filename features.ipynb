```python
# ---------------------------------------------
# Liquid Conversion Tool: Gallons ↔ Pounds
# ---------------------------------------------

def convert_liquid(weight_lb_per_gal, quantity_gal):
    """
    Convert between gallons and pounds for a liquid product.

    Parameters:
        weight_lb_per_gal (float): Weight of the liquid in pounds per gallon.
        quantity_gal (float): Quantity of liquid in gallons.

    Returns:
        tuple: (total_pounds, total_gallons)
    """

    total_pounds = weight_lb_per_gal * quantity_gal
    total_gallons = total_pounds / weight_lb_per_gal

    return total_pounds, total_gallons


# ------------- MAIN PROGRAM -------------
if __name__ == "__main__":
    print("🔹 Liquid Conversion Tool (Gallons ↔ Pounds)")
    print("--------------------------------------------")

    # Get user inputs
    try:
        weight = float(input("Enter the weight (lb/gal): "))
        quantity = float(input("Enter the quantity (gallons): "))

        pounds, gallons = convert_liquid(weight, quantity)

        print("\n✅ Conversion Results:")
        print(f"Total Pounds : {pounds:,.2f} lb")
        print(f"Total Gallons: {gallons:,.2f} gal")

    except ValueError:
        print("❌ Invalid input. Please enter numeric values.")

convert_liquid(5,5)
```
